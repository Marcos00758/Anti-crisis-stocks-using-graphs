```python
# Pega os valores do site do Yahoo Finance
import yfinance as yf

import pandas as pd
import re
import numpy as np
import requests
```

# Dados das Ações

Colocou-se as ações mais relevantes da bolsa americana um dataframe com os dados díarios (preço de abertura e fechamento) das ações desejadas no período escolhido

```python
def pegar_todas_acoes_bolsa_br():
    consulta = yf.EquityQuery('and', [
        yf.EquityQuery('eq', ['region', 'br']),
        yf.EquityQuery('eq', ['exchange', 'SAO'])
    ])

    resultados = []
    posicao_item = 0

    while True:
        lote = yf.screen(consulta, offset=posicao_item, size=250,
                        sortField='ticker', sortAsc=True)['quotes']
        if not lote:
            break
        resultados += lote
        posicao_item += len(lote)

    # Pega só ações, retirando ativos como fundos imobiliários.
    lista_acoes = sorted({
        a['symbol'].replace('.SA', '') + '.SA'
        for a in resultados
        if re.fullmatch(r'[A-Z]{4}[3-8](\.SA)?', a['symbol'])
    })
    
    return lista_acoes

acoes = pegar_todas_acoes_bolsa_br()

# Função para baixar em lotes de 100 em 100 para não dar problema na requisição.
def baixar_acoes(acoes:list,data_inicial:str, data_final:str, partes=[]):
    for i in range(0, len(acoes), 100):
        dados = yf.download(acoes[i:i + 100], start=data_inicial, end=data_final,
                            auto_adjust=True, progress=False)
        partes.append(dados)
    
    df = pd.concat(partes, axis=1).dropna(axis=1, how='all')
    
    return df

df = baixar_acoes(acoes,"2008-01-01","2026-01-01")
```

# Tabela com a variação diária percentual de cada ação

```python
# Variação diária do preço da ação (Valor de ontem - Valor de hoje)
variacao_diaria = df.pct_change(fill_method=None) * 100

variacao_diaria = variacao_diaria['Close']
variacao_diaria
```

# Tabela com Retornos diários logaritmicos
Calcula o retorno logaritmo do preço de t / preço de (t-1)

```python
def variacao_logaritma(df):
    variacao_logaritma = np.log(df.pct_change(fill_method=None)+1)

    # Trata os erros em que um dia teve preço 0, portanto teria divisão por 0
    variacao_logaritma = variacao_logaritma.replace([np.inf, -np.inf], np.nan)
    
    return variacao_logaritma

variacao_logaritma = variacao_logaritima(df)
```

# Inflação
Tabela que calcula o valor da Ação corrigido pela inflação

```python
# IPCA mensal (%) pela API do Banco Central

def inflacao_mensal(data_inicio, data_fim, data_frame_utilizado):
       url = ("https://api.bcb.gov.br/dados/serie/bcdata.sgs.433/dados"
              f"?formato=json&dataInicial={data_inicio}&dataFinal={data_fim}")

       ipca = pd.DataFrame(requests.get(url).json())
       ipca['data'] = pd.to_datetime(ipca['data'], format='%d/%m/%Y')
       ipca['valor'] = ipca['valor'].astype(float)

       # Dá o valor em decimal
       ipca = ipca.set_index('data')['valor'] / 100

       # Índice acumulado
       indice_ipca = (1 + ipca).cumprod()

       # Ajuste para distribuir igualmente a inflação mensal para cada dia do mês
       indice_diario = indice_ipca.reindex(data_frame_utilizado.index, method='ffill')

       # Normaliza para que o último valor seja 1: preços em reais de hoje (fim de 2025)
       indice_diario = indice_diario / indice_diario.iloc[-1]

       # Preço real = preço nominal / índice
       preco_real = df['Close'].div(indice_diario, axis=0)

       return preco_real

preco_real = inflacao_mensal('01/01/2008', '31/12/2025', df)
preco_real
```
