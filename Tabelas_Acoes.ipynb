```python
# Pega os valores do site do Yahoo Finance
import yfinance as yf

import pandas as pd
from pandas_datareader import data as pdr
```

# Dados das Ações

Colocou-se as ações mais relevantes da bolsa americana um dataframe com os dados díarios (preço de abertura e fechamento) das ações desejadas no período escolhido

```python
# Ações mais relevantes da bolsa americana
empresas_relevantes_bolsa_americana = ["AAPL", "MSFT", "NVDA", "GOOGL", "AMZN", "META", "AVGO", "TSLA", "ORCL",
       "AMD","BRK-B", "JPM", "V", "MA", "BAC", "LLY", "UNH", "JNJ", "WMT", "COST", "PG", "KO", "XOM", "CVX"
       ]

# Baixar os dados diários
periodo_inicial = "2020-01-01"
periodo_final = "2025-12-31"
df = yf.download(empresas_relevantes_bolsa_americana,start=periodo_inicial, end=periodo_final)

df
```

# Tabela com a variação diária percentual de cada ação

```python
# Diferença de preço diária entre a ação em porcentagem
acoes_diferenca_diaria = (df['Close']-df['Open'])*100 / df['Open']
acoes_diferenca_diaria.head()
```

## Ações corrigidas pela inflação

Cria-se uma coluna com a inflação mensal.

```python
inflacao_mensal = pdr.DataReader("CPIAUCSL", "fred", start=periodo_inicial)["CPIAUCSL"]
inflacao = inflacao_mensal.iloc[-1] / inflacao_mensal

# Corrigir a data para ANO / MÊS / DIA

inflacao
```

Correção distribuindo a inflação linearmente para cada dia do mês

```python
# Pegar todos os dias 
dias = pd.date_range(inflacao.index.min(), inflacao.index.max(), freq="D")

# Distribuindo a inflação linearmente para cada dia do mês
fator_calendario = inflacao.reindex(dias).interpolate(method="time")

# Seleciona os dias em que a bolsa estava aberta (desconsidera, por exemplo, feriados em que a bolsa não abriu)
fator_diario_linear = fator_calendario.reindex(df.index)
```

# Tabela com o valor da Ação corrigido pela inflação

```python
preco_acao_corrigido = pd.concat(
    {
        "Open": df["Open"].mul(fator_diario_linear, axis=0),
        "Close": df["Close"].mul(fator_diario_linear, axis=0),
    },
    axis=1,
)

preco_acao_corrigido.head()
```
